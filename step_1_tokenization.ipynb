Given an input space $X$ if we have a sequence of $L$ inputs $(x_0, x_1, \dots, x_{L-1}), x_i \in X \ \forall \ i = 0, \dots, L-1$, we want to perform operations like
* predicting the next input i.e. find the conditional probability $P(x_L| x_0, \dots, x_{L-1})$
* predicting a missing input i.e. find the conditional probability $P(x_j | x_0, \dots, x_{j - 1}, x_{j+1}, \dots, x_{L-1})$

In natural languages, like English, we deal with sentences which are sequences of characters found on a usual QWERTY keyboard i.e. letters, digits, punctuations, symbols, etc. The `example_sequences` list given below contains a few examples.

```python
example_sequences = ["Hello world!",
                     "ABC easy as 123",
                     "The JaBBErwocky has been transformed!"
                     "How have you been?",
                     "aldjflasjdADSFDSFSA!@#@!$#% &(&*&())"]
```

# What is Tokenization?

We first have to break down a sentence into a sequence of smaller units called **tokens** using some set of rules. The set of all possible tokens in a natural language resulting from this process is called the **vocabulary** associated with those set of rules. We denote the size of the vocabulary by $N_\mathrm{vocab}$. Every member of the vocabulary is then indexed by the set $\{0, \dots, N_\mathrm{vocab} - 1\}$. This index is called the **token ID** of that token. Given a sentence, we can then define a map called a **tokenizer** which maps a sentence to a sequence of token IDs. This entire process of going from a sentence to a sequence of token IDs is called **tokenization**.

```python
from transformers import AutoTokenizer # AutoTokenizer automatically gets the Tokenizer associated with a language model
```

# Types of Tokenizations

## Character-based tokenization

In this simplest of schemes, each character is treated as a token. In fact, this is so trivial that in practice, language models that use this are said to be token-free.

```python
char_tokenizer = AutoTokenizer.from_pretrained("google/byt5-small") # ByT5 uses a character-based tokenizer
```

```python
# stopping at generating tokens
for seq in example_sequences:
    print(char_tokenizer.tokenize(seq))
```

```python
# going all the way and generating sequences of token ids
for seq in example_sequences:
    print(char_tokenizer.encode(seq))
```

```python
# checking out the vocabulary size
vocab_size = char_tokenizer.vocab_size
print(f"Vocabulary size of {char_tokenizer.name_or_path} tokenizer is {vocab_size}\n")

# printing out all the elements of the vocabulary
for i in range(vocab_size):
    print(repr(char_tokenizer.decode(i)), end=' ') # repr is used to print control characters which cannot be printed without using repr
```

## Word-based tokenization

In this case, the tokens are words which make up the sentence. Very few models (if any, at least on HuggingFace) use this strategy for reasons we will see below

## Subword tokenization

Words in a natural language have a lot of structure. Each word has a root which gets modified based on the context like 'heat' becomes 'heated' or 'preheat'. Also, some sequences of characters like 'ing', 'pre', 'ed', 'pro', etc are not words in their own right but appear very frequently because they are common prefixes and suffixes. To capture this, we allow breaking words into simpler units. These are the (kind of) tokens which are talked about in the literature.

### Example 1

```python
BERT_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased") # BERT uses a subword tokenization strategy called WordPiece
```

```python
# just generating the tokens
for seq in example_sequences:
    print(BERT_tokenizer.tokenize(seq))
```

Remarks:
* the double hash symbols indicate that the token was part of a natural language word along with the previous token
* all the capitals letters are lowercased as part of the pretokenization procedure by this model

```python
# generating the token ids
for seq in example_sequences:
    print(BERT_tokenizer.encode(seq))
```

Note that BERT tokenizer adds token IDs 101 and 102 at the beginning and the end of every sequence. Let's see which tokens these IDs refer to.

```python
BERT_tokenizer.decode([101, 102])
```

The [CLS] token stands for classification. We will see later that it captures the global properties of sequences which can then be used to classify them.

The [SEP] token stands for seperator. It just denotes that the sequence has ended.

```python
from random import sample
random_token_ids = sample(range(vocab_size), 100)
```

```python
# checking out the size of the vocabulary
vocab_size = BERT_tokenizer.vocab_size
print(f"Vocabulary size of {BERT_tokenizer.name_or_path} tokenizer is {vocab_size}\n")

# printing out the tokens associated with random token ids
for i in random_token_ids:
    print(repr(BERT_tokenizer.decode(i)), end=' ')
```

### Example 2

```python
BPE_tokenizer = AutoTokenizer.from_pretrained("gpt2") # GPT-2 uses a subword tokenization strategy called Byte-Pair Encoding (BPE)
```

```python
# just generating the tokens
for seq in example_sequences:
    print(BPE_tokenizer.tokenize(seq))
```

```python
# generating the token ids
for seq in example_sequences:
    print(BPE_tokenizer.encode(seq))
```

```python
# checking out the size of the vocabulary
vocab_size = BPE_tokenizer.vocab_size
print(f"Vocabulary size of {BPE_tokenizer.name_or_path} tokenizer is {vocab_size}\n")

# printing out the tokens associated with random token ids (but the same ones which were used for BERT_tokenizer earlier)
for i in random_token_ids:
    print(repr(BPE_tokenizer.decode(i)), end=' ')
```

# Why most language models use subword tokenization?

### Character-based

Pros:
* Simple to implement
* Small vocabulary size
* Almost no need to train the tokenizer
* Can handle any sequence of characters

Cons:
* Poor performance in standard tasks down the line - characters capture no semantics or grammer which means that for the next or the missing token, many tokens (i.e. characters) will have the same probability leading to high error rates
* Slow to train neural networks with - the length of the sequence post-tokenization is the same as before i.e. doesn't reduce the length. The training costs for transformers scale quadratically with the length of the input sequences (due to the self-attention part)

### Word-based

Pros:
* Simple to implement - just break sentences at spaces, punctuations, symbols, etc (but not letters)
* Almost no need to train the tokenizer
* Captures the semantics of the language quite well - words are the smallest units which carry meaning in most natural languages
* Faster to train compared to character-based tokenization - post-tokenization sequence length can be orders of magnitude smaller than the input sentence

Cons:
* Huge vocabulary - from hundreds of thousands to millions of tokens
* Breaks down even when the number of errors is small
* Words with same root like 'run' and 'running' are treated as being completely different

### Subword

Pros:
* Vocabulary which is orders of magnitude smaller than the number of words in a natural language
* Great performance - almost all modern LLMs use subword tokenizers
* Each token has semantic importance
* Words with the same root word are tokenized in a way which extracts the root words - this becomes really helpful downstream

Cons:
* Difficult to implement - how to identify valid subwords which are semantically rich
* Needs a lot of training to become useful

Only in the case of subword tokenization, the cons are not conceptual. We can throw in more resources to overcome them. This is why it is standard approach.
